```python
"""
This can be used to test the Introvy MCP tool server.
The premise here is to test a Python client using the asyncio library along with the Client class from fastmcp to invoke the tool by name (sorta loose coupling). The MCP will handle serialization to get the info.
"""

import asyncio
from fastmcp import Client

async def main():
    async with Client("http://gofastmcp.com/mcp") as client:
        result = await client.call_tool(
            name="SearchFastMcp",
            arguments={"query": "deploy a FastMCP server"}
        )
        print(result)
        asyncio.run(main())
```
